```python
from abc import ABC, abstractmethod    #import tools for abstract classes
class Animal(ABC):                     #Create an abstract parent class

  @abstractmethod                      #Require child classes to implement sound()
  def sound(self):             #Define the required method
    pass                       #No actual animal sound is defined here

class Dog(Animal):             #Dog inherits the rule from Animal
  def eat(self):             #Dog provides its version of sound()
    print("Dog is eating")             #Dog's implementation

class Cat(Animal):               #Cat inherits the rule from Animal
  def sound(self):                 #Cat provides its version of sound()
    print("Meow!")           #Cat's implementation

dog = Dog()       #Create a Dog object
cat = Cat()         #Create a Cat object
dog.sound()      #Ask Dog to make its sound
cat.sound()          #Ask Cat to make its sound
```

```python
from abc import ABC, abstractmethod  #import abstraction tools
class Shape(ABC):                    #Create a abstract parent called Shape
  @abstractmethod                       #Required every concrete shape to implement area()
  def area(self):                       #Define the common operation:area
    pass                           #Shape does not choose one formula

class Circle(Shape):                  #Circle follows the shape rule
  def area(self):                   #Circle provides its own area implementation
    print("I calculate the area of a circle")              #Circle's behavior

class Square(Shape):                  #Square follows the shape rule
  def area(self):                      #Square provides its own area implementation
    print("I calculate the area of asquare")        #Square's behavior

circle = Circle()                   #Create a Circle object
square = Square()                   #Create a Square object
circle.area()                      #Use Circle's implementation
square.area()                      #Use Square's implementation
```

```python
from abc import ABC, abstractmethod  #import abstraction tools
class Shape(ABC):                    #Shape is an abstract base class
  @abstractmethod                   #Every concerte Shape must implement area()
  def area(self):                       #Define the required operation
    pass                           #No one formula fits every shape

class Circle(Shape):                  #Circle inherits from Shape
  def __init__(self, radius):                   #Receive the circle's radius
    self.radius = radius            #Store the radius
  def area(self):
    return 3.14 * self.radius * self.radius   # pi x r x r
class Square(Shape):                  #Square inherits from Shape
  def __init__(self, side):                      #Receive the square's side length
    self.side = side         #Store the side length
  def area(self):
    return self.side * self.side     #Square's implementation of area()

circle = Circle(5)                   #Create a Circle with radius 5
square = Square(4)                   #Create a Square with side 4
print(circle.area())                     #Calculate and print Circle's area
print(square.area())                     #Calculate and print Square's area
```
